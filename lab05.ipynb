```python
lccn_1 = '98508155'
lccn_2 = '2017843202'
lccn_3 = '2020722343'
```

```python
import json
import requests
from pathlib import Path

baseURL = 'http://www.loc.gov/'
itemEndpoint = 'item/'
lccns = [lccn_1, lccn_2, lccn_3]
```

```python
parameters = {
    'fo' : 'json'
}
```

```python

for lccn in lccns:
    r = requests.get(baseURL + itemEndpoint + lccn, params = parameters)
    print(r.url, r.status_code)

    def jprint(r):
        text = json.dumps(r, sort_keys=True, indent=4)
    jprint

    outputs = Path("./loc_outputs")
    file_path = outputs/f"{lccn}.json"
    with file_path.open('w', encoding="utf-8") as f:
        json.dump(r.json(), f)
```
